# imports

```python
import networkx as nx
```

# show nodes/edges mapping of a graph is not preserved when going through adjacency matrix

```python
g1 = nx.star_graph(6)

g1.remove_nodes_from([4])
g1.add_nodes_from([4])

print(g1.nodes)
print(g1.edges)

A1 = nx.to_numpy_array(g1)
g2 = nx.from_numpy_array(A1)
print(g2.nodes)
print(g2.edges)
```

# show line graph preserves the original graph edges order

```python
g1 = nx.star_graph(6)

g1.remove_nodes_from([4])
g1.add_nodes_from([4])
g1.add_edges_from([(0, 4)])

g1_line = nx.line_graph(g1)

print(g1.edges)
print(g1_line.nodes)
```
